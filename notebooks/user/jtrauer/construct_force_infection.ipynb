Illustration of how to construct the force of infection
(frequency-dependent) without using `summer2`'s built-in
`add_infection_frequency_flow`.
(I think this is correct, but may need David to confirm
and haven't tested in detail.)
Also, note that this is not an epidemiologically sensible model.

```python
from summer2 import CompartmentalModel
from summer2.parameters import Parameter, Function, CompartmentValues
```

```python
comps = ["susceptible", "infectious", "recovered"]
epi_model = CompartmentalModel([0.0, 100.0], comps, ["infectious"])
```

```python
def get_force_infection(comp_vals):
    """Manually construct the force of infection calculation.
    """
    infect_comps = epi_model.query_compartments({"name": "infectious"}, as_idx=True)
    infect_pop = comp_vals[infect_comps]
    all_comps = epi_model.query_compartments({"name": comps}, as_idx=True)
    total_pop = comp_vals[all_comps].sum()
    return infect_pop / total_pop

epi_model.add_transition_flow(
    "infection",    
    Function(get_force_infection, [CompartmentValues]) * Parameter("beta"),
    "susceptible",
    "infectious",
)
```

```python
epi_model.set_initial_population({"susceptible": 100.0})
epi_model.run({"beta": 1.0})
```
